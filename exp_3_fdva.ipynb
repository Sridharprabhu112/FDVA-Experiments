```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
df = pd.read_csv("student_data.csv")
print(df)
```

```python
df.info()
df.describe(include="all")
```

```python
df.isnull().sum()
```

```python
print(f"Mean of math_score: {df['Math'].mean()}")
print(f"Median of English_score: {df['English'].median()}")
```

```python
dupli = df.duplicated()
print(dupli)
```

```python
df = df.drop_duplicates()
print(df)
```

```python
df["Math"] = df["Math"].astype(int)
df["English"] = df["English"].round(1)
print(df.dtypes)
```

```python
minmax = MinMaxScaler()

df[["Math"]] = minmax.fit_transform(df[["Math"]])

print(df)
```

```python
standard = StandardScaler()
df[["English"]] = standard.fit_transform(df[["English"]])
print(df)
```
