```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.read_csv('test.csv')

df.head()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(df['ApplicantIncome'].head(20))

plt.xlabel('Applicant Index')
plt.ylabel('Applicant Income')
plt.title('Line Chart of Applicant Income')

plt.show()
```

```python
property_counts = df['Property_Area'].value_counts()

plt.figure(figsize=(7, 5))

plt.bar(property_counts.index, property_counts.values)

plt.xlabel('Property Area')
plt.ylabel('Number of Applicants')
plt.title('Applicants by Property Area')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df['ApplicantIncome'],
        bins=20,
            edgecolor='black'
            )

plt.xlabel('Applicant Income')
plt.ylabel('Frequency')
plt.title('Distribution of Applicant Income')

plt.show()
```
